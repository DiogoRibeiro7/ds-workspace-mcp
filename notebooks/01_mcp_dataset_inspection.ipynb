# Dataset Inspection Walkthrough

This notebook demonstrates the core local inspection workflow exposed by `ds-workspace-mcp`.
It uses the bundled sample clinic dataset and focuses on the same bounded outputs an MCP client would consume.

## Goals

- discover datasets in the safe data directory;
- preview rows before deeper work;
- build a structured dataset profile;
- surface conservative data-quality issues;
- inspect basic numeric relationships;
- translate raw outputs into analyst actions.

```python
from pprint import pprint

from ds_workspace_mcp.core import (
    detect_csv_dataset_issues,
    list_csv_files,
    preview_csv_dataset,
    profile_csv_dataset,
)
from ds_workspace_mcp.diagnostics import summarize_correlations_dataset
```

```python
datasets = list_csv_files()
datasets
```

The sample dataset is intentionally small, which makes it useful for fast inspection and MCP prompt demos.
The next step is to confirm the row-level structure before reasoning about the schema.

```python
file_name = "sample_clinic_usage.csv"
preview = preview_csv_dataset(file_name=file_name, rows=5)
preview.rows
```

```python
profile = profile_csv_dataset(file_name)

summary = {
    "row_count": profile.row_count,
    "column_count": profile.column_count,
    "columns": profile.columns,
    "dtypes": profile.dtypes,
    "missing_values": profile.missing_values,
}
pprint(summary)
```

The profile output is intentionally compact. The main questions at this stage are:

- whether the time column is parseable later;
- whether any numeric fields are unexpectedly sparse;
- whether the dataset looks ready for either descriptive analysis or forecasting setup.

```python
issues = detect_csv_dataset_issues(file_name)
issues
```

```python
correlations = summarize_correlations_dataset(file_name)
correlations.top_correlations
```

## Interpretation

Typical takeaways from this sample are:

- `appointments`, `scheduled_staff`, and `avg_wait_minutes` move together, which is plausible operationally;
- the issue detector should remain quiet on a clean small sample, which is a useful baseline for later noisy synthetic datasets;
- the profile is enough to draft an assistant prompt or choose a first target variable.

For forecasting or staffing analysis, the next logical step is to validate the time-series structure on a larger dataset.
